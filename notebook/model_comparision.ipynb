```python
import pandas as pd
import joblib
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix
```

```python
data = joblib.load("../data/processed/model_data.joblib")
X_test = data["X_test"]
y_test=data['y_test']
```

```python
dt_model = joblib.load("../models/decision_tree.joblib")
lr_model = joblib.load("../models/logistic_regression.joblib")
```

```python
print("test data: ", X_test.shape)
print("decision tree loaded: ",type(dt_model).__name__)
print("logistic regression loaded:",type(lr_model).__name__)
```

```python
## comparision
```

```python
models = {"Decision Tree": dt_model, "Logistic Regression": lr_model}

results=[]
for name, model in models.items():
    y_pred=model.predict(X_test)
    y_prob=model.predict_proba(X_test)[:,1]
    accuracy=accuracy_score(y_test,y_pred)
    precision=precision_score(y_test,y_pred)
    recall=recall_score(y_test,y_pred)
    f1 = f1_score(y_test, y_pred)
    roc = roc_auc_score(y_test, y_prob)
    avg_precision=average_precision_score(y_test, y_prob)
    
    results.append([name, accuracy, precision, recall, f1, roc,avg_precision])

```

```python
columns = ["Model","Accuracy","Precision","Recall","F1 score", "ROC-AUC","Average precision"]
comparision_df = pd.DataFrame(results, columns=columns)
display(comparision_df)
```
